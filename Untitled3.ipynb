### DATA ANALYSIS

```python
# Quality Measures (Numerical Scores)
quality_features = [
    'Aroma', 'Flavor', 'Aftertaste', 'Acidity', 'Body',
    'Balance', 'Sweetness', 'Clean.Cup', 'Uniformity',
    'Cupper.Points', 'Total.Cup.Points'
]
```

```python
# Bean Metadata (Physical attributes)
bean_features = [
    'Moisture', 'Category.One.Defects', 'Category.Two.Defects',
    'Quakers', 'Number.of.Bags'
]
```

```python
# Farm Metadata (Categorical)
farm_features = [
    'Country.of.Origin', 'Region', 'Farm.Name', 'Owner',
    'Company', 'Variety', 'Processing.Method', 'Species', 'Mill'
]
```

### Univariate Analysis

```python
for col in quality_features:
    plt.figure(figsize=(6,4))
    sns.histplot(coffee_df[col], kde=True)
    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.show()
```

```python
for col in bean_features:
    plt.figure(figsize=(6,4))
    sns.histplot(coffee_df[col], kde=True,color='orange')
    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.show()
```

```python
for col in farm_features:
    plt.figure(figsize=(8,4))
    coffee_df[col].value_counts().head(10).plot(kind='bar',color='green')
    plt.title(f"Top 10 Categories - {col}")
    plt.xlabel(col)
    plt.ylabel("Count")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

```python
species_counts = coffee_df['Species'].value_counts()
plt.figure()
plt.pie(species_counts, labels=species_counts.index, autopct='%1.1f%%')
plt.title("Pie Chart: Coffee Species Distribution")
plt.show()
```

```python
high = coffee_df[coffee_df['Total.Cup.Points'] >= 85].shape[0]
medium = coffee_df[(coffee_df['Total.Cup.Points'] >= 80) & (coffee_df['Total.Cup.Points'] < 85)].shape[0]
low = coffee_df[coffee_df['Total.Cup.Points'] < 80].shape[0]

stages = ['High Quality', 'Medium Quality', 'Low Quality']
values = [high, medium, low]

plt.figure(figsize=(6,6))
plt.barh(stages, values,color='violet')
plt.gca().invert_yaxis()
plt.xlabel("Number of Samples")
plt.title("Funnel Chart of Coffee Quality Levels")
plt.show()
```

### Bivariate Analysis

```python
# Quality Measures
quality_features = [
    'Aroma', 'Flavor', 'Aftertaste', 'Acidity', 'Body',
    'Balance', 'Sweetness', 'Clean.Cup', 'Uniformity',
    'Cupper.Points', 'Total.Cup.Points'
]

# Bean Characteristics
bean_features = [
    'Moisture', 'Category.One.Defects', 
    'Category.Two.Defects', 'Quakers', 'Number.of.Bags'
]

# Farm Attributes
farm_features = [
    'Country.of.Origin', 'Variety', 'Processing.Method', 'Region'
]
```

```python
plt.figure(figsize=(12,8))
corr_matrix = coffee_df[quality_features + bean_features].corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap: Quality vs Bean Characteristics")
plt.show()
```

### Multivate Analysis

```python
x = coffee_df['Aroma']
y = coffee_df['Flavor']
target = coffee_df['Total.Cup.Points']

plt.figure(figsize=(8,6))

# Scatter plot where size represents target variable
plt.scatter(x, y, s=target,color='purple')

plt.xlabel("Aroma")
plt.ylabel("Flavor")
plt.title("Aroma vs Flavor with Total Cup Points as Target")
plt.show()
```

```python
cols = ['Aroma', 'Flavor', 'Body', 'Acidity','Aftertaste','Total.Cup.Points']
corr = coffee_df[cols].corr()

plt.figure(figsize=(8,6))
plt.imshow(corr)
plt.xticks(range(len(cols)), cols, rotation=45)
plt.yticks(range(len(cols)), cols)
plt.title("Correlation Heatmap")
plt.colorbar()
plt.show()
```

```python
plt.figure()

plt.plot(coffee_df['Cupper.Points'], coffee_df['Aroma'], label='Aroma')
plt.plot(coffee_df['Cupper.Points'], coffee_df['Flavor'], label='Flavor')
plt.plot(coffee_df['Cupper.Points'], coffee_df['Sweetness'], label='Sweetness')

plt.xlabel("Cupper Points")
plt.ylabel("Attribute Scores")
plt.title("Line Chart: Aroma, Flavor & Sweetness vs Cupper Points")
plt.legend()
plt.grid(True)
plt.show()
```
